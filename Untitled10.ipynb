```python


# 1. Check if 'has_price_record' is 100% constant (zero variance)
is_constant_price_flag = sales_long["has_price_record"].nunique() <= 1
print(f"Is 'has_price_record' constant? {is_constant_price_flag}")

# 2. Define redundant / unneeded columns to drop permanently from the DataFrame
# - wm_yr_wk: redundant with calendar columns
# - item_id_code: redundant duplicate of item_id (LightGBM natively handles 'category')
# - event_name_2, event_type_2: extremely sparse (>99.8% 'None')
cols_to_drop = ["wm_yr_wk", "item_id_code", "event_name_2", "event_type_2"]

if is_constant_price_flag:
    cols_to_drop.append("has_price_record")

sales_long.drop(columns=cols_to_drop, inplace=True, errors="ignore")
gc.collect()

# 3. Explicitly define Feature Columns (X), Target (y), and Tracking Index
# Note: 'date' is kept in the DataFrame for splitting, but excluded from features
TARGET_COL = "sales"
TRACKING_COLS = ["date"]

FEATURE_COLS = [
    col for col in sales_long.columns 
    if col not in [TARGET_COL] + TRACKING_COLS
]

print(f"Dropped columns: {cols_to_drop}\n")
print(f"Target Column (y): ['{TARGET_COL}']")
print(f"Tracking Index:   {TRACKING_COLS}\n")
print(f"Feature Columns (X) count: {len(FEATURE_COLS)}")
print(f"Feature Columns (X): {FEATURE_COLS}\n")

# 4. Inspect current DataFrame state and memory usage
sales_long.info()
```

```python

# 1. Feature Group 1: Autoregressive Lags (t-28 and t-35)
print("Generating lag_28 and lag_35...")
sales_long["lag_28"] = (
    sales_long.groupby(["store_id", "item_id"], observed=True)["sales"]
    .shift(28)
    .astype("float32")
)

sales_long["lag_35"] = (
    sales_long.groupby(["store_id", "item_id"], observed=True)["sales"]
    .shift(35)
    .astype("float32")
)
gc.collect()

# 2. Feature Group 2: Rolling Window Averages on top of lag_28
# Using lag_28 ensures zero lookahead leakage across the 28-day forecast horizon
print("Generating rolling_mean_7 and rolling_mean_28...")
grouped_lag28 = sales_long.groupby(["store_id", "item_id"], observed=True)["lag_28"]

sales_long["rolling_mean_7"] = (
    grouped_lag28.transform(lambda s: s.rolling(7, min_periods=1).mean())
    .astype("float32")
)

sales_long["rolling_mean_28"] = (
    grouped_lag28.transform(lambda s: s.rolling(28, min_periods=1).mean())
    .astype("float32")
)

del grouped_lag28
gc.collect()

# 3. Feature Group 3: Price Dynamics (Current price relative to item baseline)
print("Generating price_ratio...")
item_mean_price = (
    sales_long.groupby("item_id", observed=True)["sell_price"]
    .transform("mean")
    .astype("float32")
)
sales_long["price_ratio"] = (sales_long["sell_price"] / item_mean_price).astype("float32")

del item_mean_price
gc.collect()

print("All features computed successfully!\n")

# 4. Verification: Inspect a sample product-store timeline around Day 28
sample_item = sales_long["item_id"].iloc[0]
sample_store = sales_long["store_id"].iloc[0]

check_cols = [
    "date", "store_id", "item_id", "sales", "sell_price",
    "lag_28", "lag_35", "rolling_mean_7", "rolling_mean_28", "price_ratio"
]

sample_view = sales_long[
    (sales_long["store_id"] == sample_store) & 
    (sales_long["item_id"] == sample_item)
][check_cols]

display(sample_view.iloc[26:36])
```

```python

# 1. Update the list of active feature columns (excluding target and tracking dates)
TARGET_COL = "sales"
TRACKING_COLS = ["date"]

FEATURE_COLS = [
    col for col in sales_long.columns 
    if col not in [TARGET_COL] + TRACKING_COLS
]

print(f"Total features to be fed into models ({len(FEATURE_COLS)}):")
print(FEATURE_COLS)

# 2. Define the warmup cutoff date (first valid date with complete lag_35 history)
warmup_cutoff = sales_long["date"].min() + pd.Timedelta(days=35)

# 3. Create masks: drop the 35-day warmup from Train only
train_clean_mask = (sales_long["date"] >= warmup_cutoff) & (sales_long["date"] < val_start)
val_mask = (sales_long["date"] >= val_start) & (sales_long["date"] < test_start)
test_mask = sales_long["date"] >= test_start

# 4. Separate into Feature Matrices (X) and Target Vectors (y)
X_train = sales_long.loc[train_clean_mask, FEATURE_COLS]
y_train = sales_long.loc[train_clean_mask, TARGET_COL]

X_val = sales_long.loc[val_mask, FEATURE_COLS]
y_val = sales_long.loc[val_mask, TARGET_COL]

X_test = sales_long.loc[test_mask, FEATURE_COLS]
y_test = sales_long.loc[test_mask, TARGET_COL]

# 5. Verification
print("\n--- Final Matrix Allocations ---")
print(f"X_train: {X_train.shape} | Nulls in lags: {X_train['lag_35'].isna().sum()}")
print(f"X_val:   {X_val.shape}   | Nulls in lags: {X_val['lag_35'].isna().sum()}")
print(f"X_test:  {X_test.shape}  | Nulls in lags: {X_test['lag_35'].isna().sum()}")

# Free unreferenced variables
gc.collect()
```

```python


lag_and_rolling_cols = ["lag_28", "lag_35", "rolling_mean_7", "rolling_mean_28"]

# 1. Fill lag/rolling nulls with 0.0 (reflects zero historical velocity for new products)
for col in lag_and_rolling_cols:
    X_train[col] = X_train[col].fillna(0.0)
    X_val[col] = X_val[col].fillna(0.0)
    X_test[col] = X_test[col].fillna(0.0)

# 2. Fill price_ratio nulls with 1.0 (baseline normal price)
X_train["price_ratio"] = X_train["price_ratio"].fillna(1.0)
X_val["price_ratio"] = X_val["price_ratio"].fillna(1.0)
X_test["price_ratio"] = X_test["price_ratio"].fillna(1.0)

gc.collect()

# 3. Final verification across all matrices
print("Null count check after imputation:")
print(f"X_train remaining nulls:\n{X_train.isna().sum()[X_train.isna().sum() > 0]}")
print(f"X_val remaining nulls:\n{X_val.isna().sum()[X_val.isna().sum() > 0]}")
print(f"X_test remaining nulls:\n{X_test.isna().sum()[X_test.isna().sum() > 0]}")

print("\nAll feature matrices are 100% complete, numeric/categorical, and leak-free!")
```

```python

# Global leaderboard tracker
leaderboard = []

def evaluate_predictions(y_true, y_pred, model_name="Model"):
    # Ensure non-negative predictions for sales demand
    y_pred_clipped = np.clip(y_pred, 0, None)
    
    mae = np.mean(np.abs(y_true - y_pred_clipped))
    rmse = np.sqrt(np.mean((y_true - y_pred_clipped) ** 2))
    wape = (np.sum(np.abs(y_true - y_pred_clipped)) / np.sum(y_true)) * 100
    
    result = {
        "Model": model_name,
        "MAE": round(float(mae), 4),
        "RMSE": round(float(rmse), 4),
        "WAPE (%)": round(float(wape), 2)
    }
    leaderboard.append(result)
    return pd.DataFrame([result])
```

```python
import lightgbm as lgb

# 1. Configure the baseline model
lgb_baseline = lgb.LGBMRegressor(
    objective="regression",      # Standard mean squared error
    metric="rmse",
    n_estimators=600,            # Cap on total boosting rounds
    learning_rate=0.08,          # Stable step size
    num_leaves=63,               # Moderate tree depth capacity
    subsample=0.8,               # Row subsampling per tree
    colsample_bytree=0.8,        # Feature subsampling per tree
    random_state=42,
    n_jobs=-1                    # Utilize all CPU cores
)

# 2. Fit with Early Stopping on the Validation Set
print("Fitting LightGBM (RMSE Baseline)...")
lgb_baseline.fit(
    X_train, y_train,
    eval_set=[(X_val, y_val)],
    callbacks=[
        lgb.early_stopping(stopping_rounds=30, verbose=50),
        lgb.log_evaluation(period=50)
    ]
)

# 3. Generate predictions on the untouched Test Set
print("\nPredicting on Test Set...")
y_test_pred_lgb_base = lgb_baseline.predict(X_test)

# 4. Score against true test sales
metrics_df = evaluate_predictions(y_test, y_test_pred_lgb_base, model_name="LightGBM (RMSE Baseline)")
display(pd.DataFrame(leaderboard))

gc.collect()
```
